```python
import pandas as pd

path = "/kaggle/input/datasets/aabdollahii/humanvsai/HVA-Grok-GPT-Qwen.xlsx"
df = pd.read_excel(path)

print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
df.head(10)
```

```python
model_map = {"text": "human", "MODEL1": "qwen", "MODEL2": "grok", "MODEL3": "gpt"}

long_df = df.melt(
    id_vars=["year", "filename", "word_count"],
    value_vars=list(model_map.keys()),
    var_name="source_col",
    value_name="content",
)

long_df["source"] = long_df["source_col"].map(model_map)
long_df["label"] = (long_df["source"] != "human").astype(int)   # 0 = human, 1 = machine
long_df = long_df.drop(columns=["source_col"]).dropna(subset=["content"])
long_df["content"] = long_df["content"].astype(str).str.strip()
long_df["n_words"] = long_df["content"].str.split().str.len()

long_df = long_df.sort_values(["filename", "source"]).reset_index(drop=True)

print(long_df.shape)
print(long_df["source"].value_counts())
long_df.head(12)
```
