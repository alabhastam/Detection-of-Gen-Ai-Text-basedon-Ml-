```python
import pandas as pd

path = "/kaggle/input/datasets/aabdollahii/humanvsai/HVA-Grok-GPT-Qwen.xlsx"
df = pd.read_excel(path)

print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
df.head(10)
```

```python
model_map = {"text": "human", "MODEL1": "qwen", "MODEL2": "grok", "MODEL3": "gpt"}

long_df = df.melt(
    id_vars=["year", "filename", "word_count"],
    value_vars=list(model_map.keys()),
    var_name="source_col",
    value_name="content",
)

long_df["source"] = long_df["source_col"].map(model_map)
long_df["label"] = (long_df["source"] != "human").astype(int)   # 0 = human, 1 = machine
long_df = long_df.drop(columns=["source_col"]).dropna(subset=["content"])
long_df["content"] = long_df["content"].astype(str).str.strip()
long_df["n_words"] = long_df["content"].str.split().str.len()

long_df = long_df.sort_values(["filename", "source"]).reset_index(drop=True)

print(long_df.shape)
print(long_df["source"].value_counts())
long_df.head(12)
```

```python
import pandas as pd
import re
from hazm import Normalizer

# Load your dataset
path = "/kaggle/input/datasets/aabdollahii/humanvsai/HVA-Grok-GPT-Qwen.xlsx"
df = pd.read_excel(path)

# Initialize Hazm Normalizer
normalizer = Normalizer()

def preprocess_persian_text(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Remove newlines and excess whitespace
    text = text.replace('\n', ' ').replace('\r', ' ')
    text = re.sub(r'\s+', ' ', text).strip()
    
    # 2. Normalize using Hazm (handles Persian characters, alef/ye/ke standardization)
    text = normalizer.normalize(text)
    
    # 3. Convert all English digits to Persian digits
    english_to_persian = {
        '0': '۰', '1': '۱', '2': '۲', '3': '۳', '4': '۴',
        '5': '۵', '6': '۶', '7': '۷', '8': '۸', '9': '۹'
    }
    for eng, per in english_to_persian.items():
        text = text.replace(eng, per)
        
    return text

text_column_name = 'text' 
if text_column_name in df.columns:
    df[text_column_name] = df[text_column_name].apply(preprocess_persian_text)

# Save to output directory in Kaggle
output_path = "/kaggle/working/cleaned_HVA_dataset.xlsx"
df.to_excel(output_path, index=False)

print(f"Preprocessing complete. Saved to: {output_path}")
print(df.head(5))
```

# long version

```python

normalizer = Normalizer()

digit_map = str.maketrans({
    "0": "۰", "1": "۱", "2": "۲", "3": "۳", "4": "۴",
    "5": "۵", "6": "۶", "7": "۷", "8": "۸", "9": "۹",
    "٠": "۰", "١": "۱", "٢": "۲", "٣": "۳", "٤": "۴",
    "٥": "۵", "٦": "۶", "٧": "۷", "٨": "۸", "٩": "۹",
})

def preprocess_persian_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Remove newline-like leakage characters
    text = text.replace("\n", " ")
    text = text.replace("\r", " ")
    text = text.replace("\t", " ")
    text = text.replace("\v", " ")
    text = text.replace("\f", " ")

    # Remove escaped newline forms if they exist as literal text
    text = text.replace("\\n", " ")
    text = text.replace("\\r", " ")
    text = text.replace("\\t", " ")

    # Remove HTML/XML tags if present
    text = re.sub(r"<[^>]+>", " ", text)

    # Normalize Persian text with Hazm
    text = normalizer.normalize(text)

    # Convert Arabic/Persian variants and invisible spacing issues
    text = text.replace("ك", "ک")
    text = text.replace("ي", "ی")
    text = text.replace("ى", "ی")
    text = text.replace("ة", "ه")
    text = text.replace("ؤ", "و")
    text = text.replace("إ", "ا")
    text = text.replace("أ", "ا")
    text = text.replace("ٱ", "ا")

    # Remove Arabic diacritics
    text = re.sub(r"[\u064B-\u065F\u0670\u06D6-\u06ED]", "", text)

    # Convert all digits to Persian digits
    text = text.translate(digit_map)

    # Remove repeated whitespace again after normalization
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Apply preprocessing to the long-format dataframe
long_df["content"] = long_df["content"].apply(preprocess_persian_text)

# Remove empty rows after cleaning
long_df = long_df[long_df["content"].str.len() > 0].reset_index(drop=True)

# Recalculate simple word count after cleaning, without tokenization
long_df["n_words"] = long_df["content"].str.split().str.len()

# Save cleaned long-format dataset to Kaggle output
output_xlsx = "/kaggle/working/HVA_long_preprocessed.xlsx"
output_csv = "/kaggle/working/HVA_long_preprocessed.csv"

long_df.to_excel(output_xlsx, index=False)
long_df.to_csv(output_csv, index=False, encoding="utf-8-sig")

print("Saved Excel:", output_xlsx)
print("Saved CSV:", output_csv)
print(long_df.shape)
print(long_df["source"].value_counts())
long_df.head(12)
```
